```python
import matplotlib.pyplot as plt
matriz = evaluacion["modelo"]["matriz"]
fig, ax = plt.subplots(figsize=(5,4))
ax.imshow(matriz, cmap="Greens")
for i in range(2):
    for j in range(2): ax.text(j,i,str(matriz[i][j]),ha="center",va="center")
ax.set(xticks=[0,1],yticks=[0,1],xticklabels=["Sin indicios","Revisar"],yticklabels=["Sin indicios","Revisar"],xlabel="Predicción",ylabel="Etiqueta esperada",title="Piloto sintético · US-15")
plt.show()
```
